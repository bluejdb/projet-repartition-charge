```python
df[['Types_phases', 'ChargeTotaleActualisee']].head()
```

# conclusion 
Le test ANOVA réalisé sur la charge totale actualisée selon les différentes phases montre une différence statistiquement significative entre les moyennes des phases (F = 1301.19, p < 0.001).
Cela confirme que les phases ne consomment pas du tout la même quantité de charge.

L’hypothèse H2 — “Certaines phases consomment beaucoup plus de ressources que d’autres” — est donc validée.

# TEST 3 — Corrélation Pearson : ChargeTotaleEstimee vs ChargeTotaleActualisee

```python
df["ChargeTotaleEstimee"] = pd.to_numeric(df["ChargeTotaleEstimee"], errors="coerce")
df["ChargeTotaleActualisee"] = pd.to_numeric(df["ChargeTotaleActualisee"], errors="coerce")
```

```python
df_corr = df.dropna(subset=["ChargeTotaleEstimee", "ChargeTotaleActualisee"])
```

```python
from scipy.stats import pearsonr

r, p_value = pearsonr(df_corr["ChargeTotaleEstimee"], df_corr["ChargeTotaleActualisee"])
r, p_value
```

# conclusion
Il existe une corrélation très forte (r = 0.89) et statistiquement significative (p < 0.001) entre la charge estimée et la charge réellement exécutée.
Les estimations de charge sont globalement fiables et suivent très bien la charge réelle, même si des écarts peuvent exister au niveau des centres ou des modules.
